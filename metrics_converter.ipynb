```python
from pathlib import Path
import pandas as pd
from sklearn.metrics import accuracy_score, f1_score
```

```python
dir_path = Path("CSV")
prediction_files = [f for f in dir_path.iterdir() if f.name.endswith("_predictions.csv")]
```

```python
prediction_files
```

```python
for csv_file in prediction_files:
    df = pd.read_csv(csv_file)
    accuracy = accuracy_score(df["true_label"], df["pred_label"])
    macro_f1 = f1_score(df["true_label"], df["pred_label"], average="macro")
    print(f"File: {csv_file}")
    print(f"Accuracy: {accuracy}")
    print(f"Macro F1: {macro_f1}")
```
